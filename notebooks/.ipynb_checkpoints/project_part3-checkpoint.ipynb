# Description
I started by reading the .shp files for the precipitation areas and electricity areas and creating 1,000 points per region, following the professor’s examples. That’s when things got complicated for me. Downloading the precipitation and temperature data for all the points was very difficult due to the API’s request limits. I tried using the professor’s code, and I tried using the AI, but I was still stuck—I couldn’t download the data for more than one region. I started with NO3 as requested, so I was able to do all the work on the notebook—that is, display the average precipitation and outliers using DCT high-pass filtering, and do the same with the temperatures using LOF. Then, by reusing the data from Magasinstatistikk, we analyzed the time series of reservoir filling levels using STL to separate the observed time series into trend, seasonal, and residual components, and a spectrogram to analyze the frequency content of reservoir filling variations over time. All the plots were generated using the NO3 data from 2024, but I then tried to download data from the other regions—to no avail. So now I only have the NO1 data stored in a .parquet file, which I’m using for the Streamlit portion. I set it up as if I had multiple files available—meaning that if I manage to get the data for the other regions in .parquet format, Streamlit will work directly with them, and selection will be possible.

```python
import openmeteo_requests
import requests_cache

from retry_requests import retry

# Open-Meteo API
OPEN_METEO_URL = "https://archive-api.open-meteo.com/v1/archive"

# Setup cache
cache_session = requests_cache.CachedSession(
    ".cache",
    expire_after=-1
)

# Setup retry mechanism
retry_session = retry(
    cache_session,
    retries=5,
    backoff_factor=0.2
)

# Initialize Open-Meteo client
openmeteo = openmeteo_requests.Client(
    session=retry_session
)
```

```python
def download_weather_data(
    region_id: str,
    start_date: str,
    end_date: str,
    locations: pd.DataFrame,
    batch_size: int = 50,
) -> pd.DataFrame:

    # 1. Normalize zone names
    zone_names = (
        locations["zone"]
        .astype(str)
        .str.replace(" ", "", regex=False)
        .str.upper()
    )

    # 2. Select locations belonging to the requested region
    region_locations = locations[
        zone_names == region_id.replace(" ", "").upper()
    ].reset_index(drop=True)

    if region_locations.empty:
        raise ValueError(f"No locations found for zone {region_id}.")

    all_weather_data = []

    # 3. Process locations in batches
    for batch_start in range(0, len(region_locations), batch_size):

        batch = region_locations.iloc[
            batch_start:batch_start + batch_size
        ]

        print(
            f"Downloading points "
            f"{batch_start + 1} to "
            f"{batch_start + len(batch)} "
            f"out of {len(region_locations)}..."
        )

        # 4. Prepare multiple coordinates for one API request
        params = {
            "latitude": batch["latitude"].astype(float).tolist(),
            "longitude": batch["longitude"].astype(float).tolist(),
            "start_date": start_date,
            "end_date": end_date,
            "hourly": [
                "temperature_2m",
                "precipitation",
            ],
            "models": "era5",
            "timezone": "Europe/Oslo",
        }

        # 5. Download weather data using cache and retry
        responses = openmeteo.weather_api(
            OPEN_METEO_URL,
            params=params
        )

        # 6. Process each location returned by the API
        for (_, location), response in zip(
            batch.iterrows(),
            responses
        ):

            hourly = response.Hourly()

            # Extract hourly variables
            temperature = hourly.Variables(0).ValuesAsNumpy()
            precipitation = hourly.Variables(1).ValuesAsNumpy()

            # Generate timestamps
            timestamps = pd.date_range(
                start=pd.to_datetime(
                    hourly.Time(),
                    unit="s",
                    utc=True
                ),
                periods=len(temperature),
                freq=pd.Timedelta(seconds=hourly.Interval()),
            )

            # Create DataFrame
            weather_df = pd.DataFrame({
                "timestamp": timestamps,
                "temperature_2m": temperature,
                "precipitation": precipitation,
            })

            # Add location metadata
            weather_df["region"] = region_id.replace(" ", "").upper()
            weather_df["point_id"] = location["point_id"]
            weather_df["latitude"] = location["latitude"]
            weather_df["longitude"] = location["longitude"]

            all_weather_data.append(weather_df)

    # 7. Combine all batches
    if not all_weather_data:
        return pd.DataFrame()

    final_df = pd.concat(
        all_weather_data,
        ignore_index=True
    )

    print(
        f"\nDownload completed: "
        f"{len(final_df)} weather records retrieved."
    )

    return final_df
```

```python
import os

os.makedirs("../data", exist_ok=True)

zones = ["NO1", "NO2", "NO3", "NO4", "NO5"]

weather_by_zone = {}

for zone in zones:

    print(f"\nDownloading {zone}...")

    weather_by_zone[zone] = download_weather_data(
        region_id=zone,
        start_date="2024-01-01",
        end_date="2024-12-31",
        locations=electricity_df,
        batch_size=50,
    )

    file_path = f"../data/weather_{zone}.parquet"

    weather_by_zone[zone].to_parquet(
        file_path,
        index=False,
        compression="snappy"
    )

    print(
        f"{zone} saved: "
        f"{len(weather_by_zone[zone]):,} records | "
        f"{os.path.getsize(file_path) / (1024**2):.2f} MB"
    )
```

```python
import numpy as np
from scipy.fft import dct, idct

# Original precipitation series
precipitation = precipitation_mean["precipitation"].values

# Apply the Discrete Cosine Transform (DCT)
precipitation_dct = dct(precipitation, type=2, norm="ortho")

# Define the frequency cutoff
cutoff = 100

# Remove low-frequency components
precipitation_dct_filtered = precipitation_dct.copy()
precipitation_dct_filtered[:cutoff] = 0

# Apply the inverse DCT to obtain the high-pass filtered signal
satv = idct(
    precipitation_dct_filtered,
    type=2,
    norm="ortho"
)

print("Original signal length:", len(precipitation))
print("Filtered signal length:", len(satv))
```

```python
# Calculate robust statistics from the filtered signal (SATV)
satv_median = np.median(satv)

# Median Absolute Deviation (MAD)
satv_mad = np.median(np.abs(satv - satv_median))

print("SATV median:", satv_median)
print("SATV MAD:", satv_mad)

# Number of robust standard deviations
n_std = 3

# Convert MAD into a robust standard deviation estimate
robust_std = 1.4826 * satv_mad

# Calculate Statistical Process Control boundaries
lower_limit = satv_median - n_std * robust_std
upper_limit = satv_median + n_std * robust_std

print("Lower Control Limit:", lower_limit)
print("Upper Control Limit:", upper_limit)
```

```python
# Identify outliers based on the filtered signal
outliers = (satv < lower_limit) | (satv > upper_limit)

# Add the results to the original precipitation DataFrame
precipitation_mean["is_outlier"] = outliers

print("Number of outliers:", outliers.sum())
print("Percentage of outliers:", round(outliers.mean() * 100, 2), "%")
```

```python
def detect_precipitation_outliers(
    weather_df,
    region_id,
    cutoff=100,
    n_std=3,
):
    # 1. Calculate average precipitation for each timestamp
    precipitation_mean = (
        weather_df.groupby("timestamp")["precipitation"]
        .mean()
        .reset_index()
        .sort_values("timestamp")
    )

    # 2. Extract precipitation values
    precipitation = precipitation_mean["precipitation"].to_numpy()

    # Check that the series is valid
    if len(precipitation) == 0 or not np.isfinite(precipitation).all():
        raise ValueError("Precipitation data is empty or contains missing values.")

    if not 0 <= cutoff < len(precipitation):
        raise ValueError("cutoff must be between 0 and the number of observations.")

    if n_std <= 0:
        raise ValueError("n_std must be positive.")

    # 3. Apply DCT high-pass filtering
    precipitation_dct = dct(
        precipitation,
        type=2,
        norm="ortho"
    )

    precipitation_dct[:cutoff] = 0

    satv = idct(
        precipitation_dct,
        type=2,
        norm="ortho"
    )

    # 4. Calculate robust statistics
    satv_median = np.median(satv)
    satv_mad = np.median(np.abs(satv - satv_median))

    robust_std = 1.4826 * satv_mad

    # 5. Calculate SPC boundaries
    lower_limit = satv_median - n_std * robust_std
    upper_limit = satv_median + n_std * robust_std

    # 6. Identify outliers using SATV
    outlier_mask = (
        (satv < lower_limit) |
        (satv > upper_limit)
    )

    precipitation_mean["is_outlier"] = outlier_mask

    # 7. Extract outlier observations
    outliers_df = precipitation_mean[
        precipitation_mean["is_outlier"]
    ].copy()

    # 8. Create final plot
    figure, axis = plt.subplots(figsize=(16, 6))

    axis.plot(
        precipitation_mean["timestamp"],
        precipitation_mean["precipitation"],
        color="steelblue",
        linewidth=0.7,
        label="Average precipitation"
    )

    axis.scatter(
        outliers_df["timestamp"],
        outliers_df["precipitation"],
        color="red",
        s=12,
        label="Outliers",
        zorder=3
    )

    axis.axhline(
        upper_limit,
        color="red",
        linestyle="--",
        label="Upper control limit (UCL)"
    )

    axis.axhline(
        lower_limit,
        color="red",
        linestyle="--",
        label="Lower control limit (LCL)"
    )

    axis.set_title(
        f"Average hourly precipitation and detected outliers — "
        f"{region_id} (2024) - cutoff={cutoff}, n_std={n_std}"
    )

    axis.set_xlabel("Date")
    axis.set_ylabel("Precipitation (mm)")
    axis.grid(alpha=0.3)
    axis.legend()

    figure.tight_layout()

    # 9. Create statistical summary
    summary = {
        "region": region_id,
        "cutoff": cutoff,
        "n_std": n_std,
        "median": satv_median,
        "MAD": satv_mad,
        "lower_limit": lower_limit,
        "upper_limit": upper_limit,
        "number_of_outliers": len(outliers_df),
        "percentage_of_outliers": outlier_mask.mean() * 100,
    }

    return figure, outliers_df, summary
```

```python
from sklearn.neighbors import LocalOutlierFactor

def detect_temperature_outliers(
    weather_df,
    region_id,
    outlier_fraction=0.01,
    n_neighbors=20,
):
    # 1. Calculate average temperature for each timestamp
    temperature_mean = (
        weather_df.groupby("timestamp")["temperature_2m"]
        .mean()
        .reset_index()
        .sort_values("timestamp")
    )

    # 2. Remove missing temperature values for LOF
    valid_data = temperature_mean.dropna(
        subset=["temperature_2m"]
    ).copy()

    # Check input parameters
    if not 0 < outlier_fraction <= 0.5:
        raise ValueError("outlier_fraction must be between 0 and 0.5.")

    if len(valid_data) <= n_neighbors:
        raise ValueError("Not enough observations for the selected n_neighbors.")

    # 3. Prepare data for LOF
    X = valid_data[["temperature_2m"]]

    # 4. Initialize and apply LOF
    lof = LocalOutlierFactor(
        contamination=outlier_fraction,
        n_neighbors=n_neighbors
    )

    lof_labels = lof.fit_predict(X)

    # 5. Identify outliers
    valid_data["is_outlier"] = lof_labels == -1

    outliers_df = valid_data[
        valid_data["is_outlier"]
    ].copy()

    # 6. Create the final plot
    figure, axis = plt.subplots(figsize=(16, 6))

    axis.plot(
        temperature_mean["timestamp"],
        temperature_mean["temperature_2m"],
        color="steelblue",
        linewidth=0.7,
        label="Average temperature"
    )

    axis.scatter(
        outliers_df["timestamp"],
        outliers_df["temperature_2m"],
        color="red",
        s=15,
        label="LOF outliers",
        zorder=3
    )

    axis.set_title(
        f"Average hourly temperature and detected outliers — {region_id}"
    )
    axis.set_xlabel("Date")
    axis.set_ylabel("Temperature (°C)")
    axis.grid(alpha=0.3)
    axis.legend()

    figure.tight_layout()

    # 7. Create statistical summary
    summary = {
        "region": region_id,
        "outlier_fraction": outlier_fraction,
        "n_neighbors": n_neighbors,
        "number_of_observations": len(valid_data),
        "number_of_outliers": len(outliers_df),
        "percentage_of_outliers": (
            len(outliers_df) / len(valid_data) * 100
        ),
        "mean_outlier_temperature": (
            outliers_df["temperature_2m"].mean()
        ),
        "min_outlier_temperature": (
            outliers_df["temperature_2m"].min()
        ),
        "max_outlier_temperature": (
            outliers_df["temperature_2m"].max()
        ),
    }

    return figure, outliers_df, summary
```

```python
import requests
# NVE Magasinstatistikk API
url = "https://biapi.nve.no/magasinstatistikk/api/Magasinstatistikk/HentOffentligData"

# Send API request
response = requests.get(url, timeout=30)
response.raise_for_status()

# Convert JSON response into a DataFrame
data = response.json()
df = pd.DataFrame(data)

reservoirs = df.rename(columns={
    "dato_Id": "date_id",
    "omrType": "area_type",
    "omrnr": "area_number",
    "iso_aar": "iso_year",
    "iso_uke": "iso_week",
    "fyllingsgrad": "filling_degree",
    "kapasitet_TWh": "capacity_TWh",
    "fylling_TWh": "filling_TWh",
    "neste_Publiseringsdato": "next_publication_date",
    "fyllingsgrad_forrige_uke": "filling_degree_previous_week",
    "endring_fyllingsgrad": "change_in_filling_degree"
})

# Convert date columns to datetime
reservoirs["date_id"] = pd.to_datetime(reservoirs["date_id"])

reservoirs["next_publication_date"] = pd.to_datetime(
    reservoirs["next_publication_date"]
)

# Replace invalid year 1 dates with NaT
reservoirs.loc[
    reservoirs["next_publication_date"].dt.year == 1,
    "next_publication_date"
] = pd.NaT

# Sort observations chronologically
reservoirs = reservoirs.sort_values("date_id")

display(reservoirs.head())
```

```python
from statsmodels.tsa.seasonal import STL
def plot_stl_decomposition(
    data,
    seasonal=13,
    trend=53,
    robust=True,
    period=52
):
    # 1. Prepare the time series
    filling_series = (
        data.set_index("date_id")["filling_degree"]
        .sort_index()
    )

    # 2. Check parameters
    if seasonal < 7 or seasonal % 2 == 0:
        raise ValueError("seasonal must be an odd integer >= 7.")

    if trend <= period or trend % 2 == 0:
        raise ValueError("trend must be odd and greater than period.")

    if filling_series.isna().any():
        raise ValueError("The time series contains missing values.")

    # 3. Perform STL decomposition
    stl = STL(
        filling_series,
        period=period,
        seasonal=seasonal,
        trend=trend,
        robust=robust
    )

    stl_result = stl.fit()

    # 4. Plot decomposition
    # Create the figure with the desired size
    figure = plt.figure(figsize=(14, 9))

    # Plot the STL decomposition
    stl_result.plot()

    figure.suptitle(
        "STL decomposition — Norwegian reservoir filling degree",
        fontsize=14,
        y=1.02
    )

    figure.tight_layout()

    return figure
```

```python
# Run STL decomposition
stl_figure = plot_stl_decomposition(
    data=reservoirs_norway,
    seasonal=13,
    trend=53,
    robust=True,
    period=52
)

# Display the plot
plt.show()
```

```python
from scipy.signal import spectrogram
def plot_spectrogram(
    data,
    window_length=52,
    window_overlap=26
):
    # 1. Prepare the time series
    filling_series = (
        data.set_index("date_id")["filling_degree"]
        .sort_index()
    )

    values = filling_series.to_numpy()

    # 2. Check parameters
    if window_length <= 0 or window_length > len(values):
        raise ValueError("Invalid window_length.")

    if window_overlap < 0 or window_overlap >= window_length:
        raise ValueError(
            "window_overlap must be >= 0 and smaller than window_length."
        )

    if np.isnan(values).any():
        raise ValueError("The time series contains missing values.")

    # 3. Compute the spectrogram
    frequencies, times, power = spectrogram(
        values,
        fs=1,  # One observation per week
        nperseg=window_length,
        noverlap=window_overlap
    )

    # 4. Convert time positions into dates
    dates = (
        filling_series.index[0]
        + pd.to_timedelta(times, unit="W")
    )

    # 5. Plot the spectrogram
    figure, axis = plt.subplots(figsize=(14, 6))

    mesh = axis.pcolormesh(
        dates,
        frequencies,
        10 * np.log10(power + 1e-12),
        shading="auto",
        cmap="viridis"
    )

    axis.set_title("Spectrogram — Norwegian reservoir filling degree")
    axis.set_xlabel("Date")
    axis.set_ylabel("Frequency (cycles per week)")

    figure.colorbar(
        mesh,
        ax=axis,
        label="Power spectral density (dB)"
    )

    figure.tight_layout()

    return figure
```

```python
# Generate spectrogram
spectrogram_figure = plot_spectrogram(
    data=reservoirs_norway,
    window_length=52,
    window_overlap=26
)

# Display the plot
plt.show()
```
